```python
df = sns.load_dataset("taxis")

# 1. Check for missing values
print("Missing values in each column:")
print(df.isnull().sum())

# 2. Display columns that have missing values
print("\nColumns with missing values:")
print(df.columns[df.isnull().any()])

# 3. Fill numerical columns with median
numerical_columns = df.select_dtypes(include="number").columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

# 4. Fill categorical columns with mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# 5. Check missing values after imputation
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Convert pickup column to datetime
df["pickup"] = pd.to_datetime(df["pickup"])

# Sort the data by pickup time
df = df.sort_values("pickup")

# Plot line chart
plt.figure(figsize=(12, 6))

plt.plot(df["pickup"], df["fare"])

plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.title("Fare Over Time")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Group by pickup borough and calculate total fare
borough_fare = df.groupby("pickup_borough")["fare"].sum()

# Create bar chart
plt.figure(figsize=(10, 6))
borough_fare.plot(kind="bar")

plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.title("Total Fare by Pickup Borough")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Count trips for each payment method
payment_counts = df["payment"].value_counts()

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Trips by Payment Method")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Create histogram for distance
plt.figure(figsize=(10, 6))

plt.hist(df["distance"].dropna(), bins=20)

plt.xlabel("Distance")
plt.ylabel("Number of Trips")
plt.title("Distribution of Taxi Trip Distance")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Create box plot
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="pickup_borough",
    y="tip"
)

plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.title("Distribution of Tip Amounts by Pickup Borough")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough"
)

plt.xlabel("Distance")
plt.ylabel("Fare")
plt.title("Relationship Between Distance and Fare")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Select numerical columns
columns = ["distance", "fare", "tip", "tolls", "total"]

# Create correlation matrix
correlation_matrix = df[columns].corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Taxi Variables")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Select required columns
columns = ["distance", "fare", "tip", "total", "pickup_zone"]

# Create pair plot
sns.pairplot(
    df[columns].dropna(),
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone"
)

plt.suptitle("Pairwise Relationships of Taxi Variables by Pickup Zone", y=1.02)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Create violin plot
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="payment",
    y="fare"
)

plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.title("Distribution of Fare by Payment Method")

plt.tight_layout()
plt.show()
```
